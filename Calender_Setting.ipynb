```python
function createCalendarEventsFromSheet() {
  const sheet = SpreadsheetApp.openByUrl('https://docs.google.com/spreadsheets/d/132W7CzaAupW6AAp3KNaT229eWfZlt29FFkSubJcUIDk/edit#gid=1174787678')
                              .getSheetByName('Email');

  const calendar = CalendarApp.getCalendarById('c_d1d87c7a07ef4f40b69f8fbea7ce607beb44336dc715838f656cdb689665231c@group.calendar.google.com');
  const data = sheet.getDataRange().getValues();

  for (let i = 1; i < data.length; i++) {
    const date = data[i][0];  // Column A - Date of session
    const topic = data[i][1]; // Column B - Topic
    const email = data[i][2]; // Column C - Faculty Email
    const resource = data[i][3]; // Column D - Resource

    if (!(date instanceof Date)) continue; // skip rows with invalid date

    const title = 'Common Academic Day, Post Graduate Student, Research';
    const description = `Topic: ${topic}\nResource: ${resource}`;
    const startTime = new Date(date);
    startTime.setHours(15, 0, 0);
    const endTime = new Date(date);
    endTime.setHours(16, 0, 0);

    // Check if event already exists to prevent duplication (optional logic)
    const existingEvents = calendar.getEvents(startTime, endTime, {search: title});
    if (existingEvents.length > 0) continue;

    const event = calendar.createEvent(title, startTime, endTime, {
      description: description,
      guests: email,
      sendInvites: true
    });

    // Add email reminders
    event.removeAllReminders(); // clear default reminders
    event.addEmailReminder(7 * 24 * 60);  // 7 days before
    event.addEmailReminder(1 * 24 * 60);  // 1 day before
  }
}
```
